# Ship Mode

```python
df['Ship Mode'].value_counts()
```

```python
x = df['Ship Mode'].value_counts().index
y = df['Ship Mode'].value_counts().values
```

```python
explode = [0.1,0.1,0]
plt.rcParams['font.family'] = 'Comic Sans MS'
plt.pie(y, labels = x, explode = explode, autopct = "%1.1f%%", wedgeprops=dict(edgecolor="magenta"))
plt.legend(loc = "upper right", bbox_to_anchor = (1.2,1))
```

```python
ax = sns.countplot(x = 'Ship Mode', data = df, hue = 'Product Category')
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
ax.bar_label(ax.containers[2])
```

# sub category

```python
ax = sns.countplot(x = "Product Category", data = df[df['Product Category'] == "Office Supplies"], hue = "Product Sub-Category")
plt.legend( bbox_to_anchor = (1.2, 1))
for p in ax.patches:
    ax.annotate(
        format(p.get_height(), ','),
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha='center', va='center',
        xytext=(0, 5),  # offset
        textcoords='offset points',
        fontsize=9, color='black', weight='bold'
    )
```

```python
ax = sns.countplot(x = "Product Category", data = df[df['Product Category'] == "Technology"], hue = "Product Sub-Category")
plt.legend( bbox_to_anchor = (1.2, 1))
for p in ax.patches:
    ax.annotate(
        format(p.get_height(), ','),
        (p.get_x() + p.get_width() / 2., p.get_height()),
        ha='center', va='center',
        xytext=(0, 5),  # offset
        textcoords='offset points',
        fontsize=9, color='black', weight='bold'
    )
```
